```python
from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi
from pymongo.errors import CollectionInvalid

DATABASE_URL = "mongodb+srv://maxiesco:maxiesco@cluster0.3zdo3lp.mongodb.net/?retryWrites=true&w=majority&appName=Cluster0"

client = MongoClient(DATABASE_URL, server_api=ServerApi('1'))

try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(f"Error al conectar a MongoDB: {e}")
    exit()


db = client.Cluster0

session_schema = {
    "$jsonSchema": {
        "bsonType": "object",
        "required": ["mentor", "startup", "date", "dateTime", "duration", "topic", "mentorSigned", "startupSigned", "status"],
        "properties": {
            "mentor": {
                "bsonType": "objectId",
                "description": "must be an ObjectId and is required"
            },
            "startup": {
                "bsonType": "objectId",
                "description": "must be an ObjectId and is required"
            },
            "date": {
                "bsonType": "date",
                "description": "must be a date and is required"
            },
            "dateTime": {
                "bsonType": "date",
                "description": "must be a date and is required"
            },
            "duration": {
                "bsonType": "double",
                "description": "must be a number between 0.5 and 8 and is required",
                "minimum": 0.5,
                "maximum": 8
            },
            "topic": {
                "bsonType": "string",
                "description": "must be a string and is required"
            },
            "summary": {
                "bsonType": "string",
                "description": "must be a string"
            },
            "mentorSigned": {
                "bsonType": "object",
                "required": ["signed"],
                "properties": {
                    "signed": {
                        "bsonType": "bool",
                        "description": "must be a boolean"
                    },
                    "timestamp": {
                        "bsonType": "date",
                        "description": "must be a date"
                    }
                }
            },
            "startupSigned": {
                "bsonType": "object",
                "required": ["signed"],
                "properties": {
                    "signed": {
                        "bsonType": "bool",
                        "description": "must be a boolean"
                    },
                    "timestamp": {
                        "bsonType": "date",
                        "description": "must be a date"
                    }
                }
            },
            "status": {
                "bsonType": "string",
                "enum": ["pending", "signed", "conflict"],
                "description": "can only be one of the enum values"
            },
            "comments": {
                "bsonType": "array",
                "items": { "bsonType": "string" },
                "description": "must be an array of strings"
            },
            "pdfUrl": {
                "bsonType": "string",
                "description": "must be a string"
            }
        }
    }
}

try:
    db.create_collection(
        "sessions",
        validator=session_schema,
        validationAction="error",
        validationLevel="strict"
    )
    print("Colección 'sessions' creada con éxito y validación de esquema aplicada.")
except CollectionInvalid:
    print("La colección 'sessions' ya existe. Si necesitas actualizar el esquema, deberás usar db.command('collMod').")
except Exception as e:
    print(f"Error al crear la colección 'sessions': {e}")

client.close()
```
