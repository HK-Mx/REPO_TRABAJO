```python

cargo_perfiles = []
email_perfiles = []
empresa_perfiles = []
foto_perfiles = []
linkedin_perfiles = []

url = "https://startupvalencia2025.notion.site/1aecf8b7c83c819fac2ddac248591626?v=1afcf8b7c83c8014a3af000cb19ea47f"

driver = webdriver.Edge()
driver.get(url)


time.sleep(5) 

sopa = bs(driver.page_source, "lxml")

# Por ejemplo, si cada perfil es un div con la clase 'profile-card':
perfiles = sopa.select("div.notion-collection-card") 

# Si el raspado requiere hacer clic en cada perfil para ver los detalles, 
# la lógica dentro del bucle deberá cambiar para interactuar con Selenium.

for i, perfil in enumerate(perfiles):
    # Aquí es donde la dinámica del código necesita un cambio fundamental
    # si los datos no están directamente en la vista principal y requieren un click.
    # Necesitamos interactuar con el driver de Selenium directamente.

    # Re-obtener los elementos 'clicables' después de cada iteración si se va a hacer click,
    # ya que los elementos del DOM pueden volverse "stale" (caducados) después de navegar.
    # O, si se hace click y se vuelve, los elementos pueden no ser los mismos.
    # Por simplicidad, asumiré que los enlaces a los perfiles están disponibles
    # y los datos internos son accesibles una vez que se hace click.

    try:
        # Intenta encontrar el enlace al perfil (si existe)
        # Esto es un selector de ejemplo, ajusta según la estructura real de Notion
        enlace_perfil_element = driver.find_elements('css selector', 'div.notion-collection-card a')[i]
        
        # Haz click en el enlace del perfil
        enlace_perfil_element.click()
        
        # Espera un tiempo para que la página de detalles del perfil cargue
        time.sleep(3) 
        
        # Ahora, raspa los detalles de la página del perfil abierta
        sopa_detalle = bs(driver.page_source, "lxml")

        # Selectores hipotéticos para los detalles del perfil.
        # ¡DEBES AJUSTAR ESTOS SELECTORES SEGÚN LA ESTRUCTURA HTML DE LA PÁGINA DE DETALLE DE NOTION!
        
        # Ejemplo: Cargo (buscando un h1 o un div específico)
        cargo = sopa_detalle.select_one("div.notion-property-text[data-property-name='Cargo'] .notion-selectable")
        cargo_perfiles.append(cargo.get_text().strip() if cargo else None)

        # Ejemplo: Email (buscando un enlace mailto o un div específico)
        email = sopa_detalle.select_one("div.notion-property-email[data-property-name='Email'] a")
        email_perfiles.append(email["href"].replace("mailto:", "").strip() if email and "href" in email.attrs else None)

        # Ejemplo: Empresa (buscando un div específico)
        empresa = sopa_detalle.select_one("div.notion-property-text[data-property-name='Empresa'] .notion-selectable")
        empresa_perfiles.append(empresa.get_text().strip() if empresa else None)

        # Ejemplo: Foto (buscando una etiqueta img)
        foto = sopa_detalle.select_one("div.notion-property-file[data-property-name='Foto'] img")
        foto_perfiles.append(foto["src"].strip() if foto and "src" in foto.attrs else None)
        
        # Ejemplo: LinkedIn (buscando un enlace a linkedin)
        linkedin = sopa_detalle.select_one("div.notion-property-url[data-property-name='LinkedIn'] a")
        linkedin_perfiles.append(linkedin["href"].strip() if linkedin and "href" in linkedin.attrs else None)

        # Vuelve a la página anterior (si la navegación te ha llevado a una nueva URL)
        driver.back()
        time.sleep(2) # Espera a que la página principal vuelva a cargar
        
    except Exception as e:
        print(f"Error al procesar el perfil {i}: {e}")
        # Si hay un error, añade None para mantener la longitud de las listas
        cargo_perfiles.append(None)
        email_perfiles.append(None)
        empresa_perfiles.append(None)
        foto_perfiles.append(None)
        linkedin_perfiles.append(None)
        # Intenta volver si el error ocurrió después de hacer click y navegar
        try:
            driver.back()
            time.sleep(2)
        except:
            pass # Si no se puede volver, simplemente continúa

driver.quit()

# Crear Series de Pandas
cargo_perfiles_series = pd.Series(cargo_perfiles)
email_perfiles_series = pd.Series(email_perfiles)
empresa_perfiles_series = pd.Series(empresa_perfiles)
foto_perfiles_series = pd.Series(foto_perfiles)
linkedin_perfiles_series = pd.Series(linkedin_perfiles)

# Crear DataFrame
ejercicio_selen_df = pd.DataFrame({
    "cargo": cargo_perfiles_series,
    "email": email_perfiles_series,
    "empresa": empresa_perfiles_series,
    "foto": foto_perfiles_series,
    "linkedin": linkedin_perfiles_series
})

print(ejercicio_selen_df)
```
